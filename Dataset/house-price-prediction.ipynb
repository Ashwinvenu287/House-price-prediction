## Corrected preprocessing and validation

This section replaces the earlier manual preprocessing/model-comparison cells. Imputation statistics are learned from training data only for the final train/test transform; cross-validation uses the same preprocessing inside each fold to avoid validation-fold leakage.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.model_selection import KFold, cross_validate
from sklearn.metrics import make_scorer, mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.compose import TransformedTargetRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

DATA_DIR = Path.cwd()
if not (DATA_DIR / "train.csv").exists():
    DATA_DIR = DATA_DIR / "Dataset"

df_train = pd.read_csv(DATA_DIR / "train.csv")
df_test = pd.read_csv(DATA_DIR / "test.csv")
y_train = df_train["SalePrice"].copy()
raw_X_train = df_train.drop(columns="SalePrice")
raw_X_test = df_test.copy()
print(f"Raw train shape: {df_train.shape}")
print(f"Raw test shape: {df_test.shape}")
```

```python
# Missingness has feature-specific meaning. These categorical NA values mean that the
# corresponding physical feature is absent, so preserve that information as "NA".
absence_categoricals = [
    "Alley", "MasVnrType", "BsmtQual", "BsmtCond", "BsmtExposure",
    "BsmtFinType1", "BsmtFinType2", "FireplaceQu", "GarageType",
    "GarageFinish", "GarageQual", "GarageCond", "PoolQC", "Fence", "MiscFeature",
]

# These ordered columns use the documented category hierarchy. Nominal columns are
# deliberately excluded and will be one-hot encoded below.
ordinal_categories = {
    "ExterQual": ["Po", "Fa", "TA", "Gd", "Ex"],
    "ExterCond": ["Po", "Fa", "TA", "Gd", "Ex"],
    "BsmtQual": ["NA", "Po", "Fa", "TA", "Gd", "Ex"],
    "BsmtCond": ["NA", "Po", "Fa", "TA", "Gd", "Ex"],
    "BsmtExposure": ["NA", "No", "Mn", "Av", "Gd"],
    "BsmtFinType1": ["NA", "Unf", "LwQ", "Rec", "BLQ", "ALQ", "GLQ"],
    "BsmtFinType2": ["NA", "Unf", "LwQ", "Rec", "BLQ", "ALQ", "GLQ"],
    "HeatingQC": ["Po", "Fa", "TA", "Gd", "Ex"],
    "KitchenQual": ["Po", "Fa", "TA", "Gd", "Ex"],
    "Functional": ["Sal", "Sev", "Maj2", "Maj1", "Mod", "Min2", "Min1", "Typ"],
    "FireplaceQu": ["NA", "Po", "Fa", "TA", "Gd", "Ex"],
    "GarageFinish": ["NA", "Unf", "RFn", "Fin"],
    "GarageQual": ["NA", "Po", "Fa", "TA", "Gd", "Ex"],
    "GarageCond": ["NA", "Po", "Fa", "TA", "Gd", "Ex"],
    "PoolQC": ["NA", "Fa", "TA", "Gd", "Ex"],
    "PavedDrive": ["N", "P", "Y"],
    # Actual Utilities categories, ordered from least to most complete service.
    "Utilities": ["ELO", "NoSeWa", "NoSewr", "AllPub"],
}

# A readable, train-only missing-value check before encoding.
def fill_for_validation(train, test):
    train, test = train.copy(), test.copy()
    for col in absence_categoricals:
        train[col] = train[col].fillna("NA")
        test[col] = test[col].fillna("NA")
    for col in train.select_dtypes(include="number").columns:
        value = train[col].median()
        train[col] = train[col].fillna(value)
        test[col] = test[col].fillna(value)
    for col in train.select_dtypes(exclude="number").columns:
        if col not in absence_categoricals:
            value = train[col].mode(dropna=True)[0]
            train[col] = train[col].fillna(value)
            test[col] = test[col].fillna(value)
    return train, test

validation_train, validation_test = fill_for_validation(raw_X_train, raw_X_test)
df = pd.concat([validation_train, validation_test], ignore_index=True)
assert df["Functional"].isna().sum() == 0
assert df["FireplaceQu"].isna().sum() == 0
assert df.isna().sum().sum() == 0
print("Functional missing values:", df["Functional"].isna().sum())
print("FireplaceQu missing values:", df["FireplaceQu"].isna().sum())
print("Remaining missing values:", df.isna().sum().sum())
print("Utilities values:", sorted(df["Utilities"].unique()))
```

```python
numeric_features = raw_X_train.select_dtypes(include="number").columns.tolist()
categorical_features = raw_X_train.select_dtypes(exclude="number").columns.tolist()
ordinal_features = list(ordinal_categories)
nominal_features = [c for c in categorical_features if c not in ordinal_features]
ordinal_absence_features = [c for c in ordinal_features if c in absence_categoricals]
ordinal_mode_features = [c for c in ordinal_features if c not in absence_categoricals]
nominal_absence_features = [c for c in nominal_features if c in absence_categoricals]
nominal_mode_features = [c for c in nominal_features if c not in absence_categoricals]

def build_preprocessor():
    return ColumnTransformer([
        ("numeric", SimpleImputer(strategy="median"), numeric_features),
        ("ordinal_absence", Pipeline([
            ("imputer", SimpleImputer(strategy="constant", fill_value="NA")),
            ("encoder", OrdinalEncoder(
                categories=[ordinal_categories[c] for c in ordinal_absence_features],
                handle_unknown="use_encoded_value", unknown_value=-1,
            )),
        ]), ordinal_absence_features),
        ("ordinal_mode", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OrdinalEncoder(
                categories=[ordinal_categories[c] for c in ordinal_mode_features],
                handle_unknown="use_encoded_value", unknown_value=-1,
            )),
        ]), ordinal_mode_features),
        ("nominal_absence", Pipeline([
            ("imputer", SimpleImputer(strategy="constant", fill_value="NA")),
            ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
        ]), nominal_absence_features),
        ("nominal_mode", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
        ]), nominal_mode_features),
    ], verbose_feature_names_out=False).set_output(transform="pandas")

# Final train/test matrices fit preprocessing on training features only.
final_preprocessor = build_preprocessor()
X_train = final_preprocessor.fit_transform(raw_X_train)
X_test = final_preprocessor.transform(raw_X_test)

# Check documented ordinal categories did not turn into unintended unknown values.
for feature, values in ordinal_categories.items():
    observed = set(df[feature].dropna().unique())
    unexpected = observed - set(values)
    assert not unexpected, f"Unexpected values in {feature}: {unexpected}"
print(f"Encoded train shape: {X_train.shape}")
print(f"Encoded test shape: {X_test.shape}")
```

```python
# Preprocessing is included inside each CV pipeline: imputation and encoding learn
# from each training fold only. Scaling is used only for scale-sensitive models.
cv = KFold(n_splits=5, shuffle=True, random_state=42)
scoring = {
    "r2": "r2",
    "mae": "neg_mean_absolute_error",
    "rmse": "neg_root_mean_squared_error",
}
models = {
    "Linear Regression": make_pipeline(build_preprocessor(), StandardScaler(), LinearRegression()),
    "SVR": TransformedTargetRegressor(
        regressor=make_pipeline(build_preprocessor(), StandardScaler(), SVR()),
        transformer=StandardScaler(),
    ),
    "Decision Tree": make_pipeline(build_preprocessor(), DecisionTreeRegressor(random_state=42)),
    "Random Forest": make_pipeline(build_preprocessor(), RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)),
    "XGBoost": make_pipeline(build_preprocessor(), XGBRegressor(random_state=42, n_jobs=-1)),
}

results = []
for name, model in models.items():
    scores = cross_validate(model, raw_X_train, y_train, cv=cv, scoring=scoring, n_jobs=1)
    results.append({
        "Model": name,
        "CV R?": scores["test_r2"].mean(),
        "CV MAE": -scores["test_mae"].mean(),
        "CV RMSE": -scores["test_rmse"].mean(),
    })
comparison = pd.DataFrame(results).set_index("Model")
print(comparison.round(3).to_string())
```

```python
# Final automated project validation
assert X_train.shape[1] == X_test.shape[1]
assert X_train.columns.equals(X_test.columns)
assert X_train.isnull().sum().sum() == 0
assert X_test.isnull().sum().sum() == 0
assert "SalePrice" not in X_test.columns
assert "SalePrice" not in X_train.columns
assert all(comparison["CV R?"] > 0), "A model has non-positive CV R?"

print("=" * 40)
print("HOUSE PRICE PROJECT VALIDATION")
print("=" * 40)
print("Missing values: PASS")
print("Train/Test alignment: PASS")
print("Target leakage: PASS")
print("Feature encoding: PASS")
print("Model training: PASS")
print("Cross-validation: PASS")
print("=" * 40)
```

## Feature engineering

Feature engineering creates a few domain-based summaries from the original housing columns. These features make age, total living area, bathrooms, and porch space explicit while keeping the transformations easy to explain. `TotalRooms` is intentionally omitted because `TotRmsAbvGrd` already includes bedrooms, so adding `BedroomAbvGr` would double-count them.

```python
import sys
import joblib
from sklearn.model_selection import RandomizedSearchCV

PROJECT_ROOT = DATA_DIR.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
from house_price_features import HousingFeatureEngineer

feature_engineer = HousingFeatureEngineer()
engineered_X_train = feature_engineer.fit_transform(raw_X_train)
engineered_X_test = feature_engineer.transform(raw_X_test)
engineered_feature_names = HousingFeatureEngineer.engineered_features

assert set(engineered_feature_names).issubset(engineered_X_train.columns)
assert set(engineered_feature_names).issubset(engineered_X_test.columns)
assert (engineered_X_train[["HouseAge", "RemodAge", "GarageAge"]].dropna() >= 0).all().all()
assert "SalePrice" not in engineered_X_train.columns
print("Engineered features:", engineered_feature_names)
print("Engineered training shape:", engineered_X_train.shape)
```

```python
# The preprocessor is rebuilt from the engineered feature table so the six new
# numeric features flow through the same leakage-safe imputation and encoding steps.
engineered_numeric_features = engineered_X_train.select_dtypes(include="number").columns.tolist()
engineered_categorical_features = engineered_X_train.select_dtypes(exclude="number").columns.tolist()
engineered_ordinal_features = list(ordinal_categories)
engineered_nominal_features = [c for c in engineered_categorical_features if c not in engineered_ordinal_features]
engineered_ordinal_absence = [c for c in engineered_ordinal_features if c in absence_categoricals]
engineered_ordinal_mode = [c for c in engineered_ordinal_features if c not in absence_categoricals]
engineered_nominal_absence = [c for c in engineered_nominal_features if c in absence_categoricals]
engineered_nominal_mode = [c for c in engineered_nominal_features if c not in absence_categoricals]

def build_engineered_preprocessor():
    return ColumnTransformer([
        ("numeric", SimpleImputer(strategy="median"), engineered_numeric_features),
        ("ordinal_absence", Pipeline([
            ("imputer", SimpleImputer(strategy="constant", fill_value="NA")),
            ("encoder", OrdinalEncoder(categories=[ordinal_categories[c] for c in engineered_ordinal_absence], handle_unknown="use_encoded_value", unknown_value=-1)),
        ]), engineered_ordinal_absence),
        ("ordinal_mode", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OrdinalEncoder(categories=[ordinal_categories[c] for c in engineered_ordinal_mode], handle_unknown="use_encoded_value", unknown_value=-1)),
        ]), engineered_ordinal_mode),
        ("nominal_absence", Pipeline([
            ("imputer", SimpleImputer(strategy="constant", fill_value="NA")),
            ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
        ]), engineered_nominal_absence),
        ("nominal_mode", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
        ]), engineered_nominal_mode),
    ], verbose_feature_names_out=False).set_output(transform="pandas")

def make_engineered_pipeline(model):
    return Pipeline([
        ("features", HousingFeatureEngineer()),
        ("preprocessor", build_engineered_preprocessor()),
        ("model", model),
    ])

def evaluate_model(name, model, X=raw_X_train):
    scores = cross_validate(model, X, y_train, cv=cv, scoring=scoring, n_jobs=1)
    return {"Model": name, "CV R2": scores["test_r2"].mean(), "CV MAE": -scores["test_mae"].mean(), "CV RMSE": -scores["test_rmse"].mean()}

engineered_models = {
    "Random Forest": make_engineered_pipeline(RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)),
    "XGBoost": make_engineered_pipeline(XGBRegressor(random_state=42, n_jobs=-1)),
}
engineered_results = pd.DataFrame([evaluate_model(name, model) for name, model in engineered_models.items()]).set_index("Model")
feature_engineering_comparison = comparison.loc[["Random Forest", "XGBoost"]].join(engineered_results, lsuffix=" Baseline", rsuffix=" Engineered")
print(feature_engineering_comparison.round(3).to_string())
```

## Beginner-friendly hyperparameter tuning

A hyperparameter is a model setting chosen before fitting, unlike a learned model parameter. `RandomizedSearchCV` tests a small number of understandable settings using the same five folds, so choices are based on validation performance rather than the Kaggle test file.

```python
# Random Forest: more trees can stabilize predictions; depth and minimum-sample settings
# control how complex each tree is. The small randomized search keeps the run practical.
rf_search = RandomizedSearchCV(
    estimator=make_engineered_pipeline(RandomForestRegressor(random_state=42, n_jobs=-1)),
    param_distributions={
        "model__n_estimators": [100, 200, 300],
        "model__max_depth": [None, 10, 20, 30],
        "model__min_samples_split": [2, 5, 10],
        "model__min_samples_leaf": [1, 2, 4],
    },
    n_iter=8, scoring="r2", cv=cv, random_state=42, n_jobs=-1, refit=True,
)
rf_search.fit(raw_X_train, y_train)
print("Random Forest best parameters:", rf_search.best_params_)
print(f"Random Forest best CV R2: {rf_search.best_score_:.3f}")

# XGBoost: estimators are boosting rounds, learning_rate is each round's step size,
# and max_depth controls the complexity of individual trees.
xgb_search = RandomizedSearchCV(
    estimator=make_engineered_pipeline(XGBRegressor(random_state=42, n_jobs=1)),
    param_distributions={
        "model__n_estimators": [100, 200, 300],
        "model__learning_rate": [0.03, 0.05, 0.1],
        "model__max_depth": [3, 5, 7],
    },
    n_iter=9, scoring="r2", cv=cv, random_state=42, n_jobs=1, refit=True,
)
xgb_search.fit(raw_X_train, y_train)
print("XGBoost best parameters:", xgb_search.best_params_)
print(f"XGBoost best CV R2: {xgb_search.best_score_:.3f}")
```

```python
tuned_rf_result = evaluate_model("Random Forest", make_engineered_pipeline(RandomForestRegressor(
    random_state=42, n_jobs=-1,
    n_estimators=rf_search.best_params_["model__n_estimators"],
    max_depth=rf_search.best_params_["model__max_depth"],
    min_samples_split=rf_search.best_params_["model__min_samples_split"],
    min_samples_leaf=rf_search.best_params_["model__min_samples_leaf"],
)))
tuned_xgb_result = evaluate_model("XGBoost", make_engineered_pipeline(XGBRegressor(
    random_state=42, n_jobs=-1,
    n_estimators=xgb_search.best_params_["model__n_estimators"],
    learning_rate=xgb_search.best_params_["model__learning_rate"],
    max_depth=xgb_search.best_params_["model__max_depth"],
)))

baseline_results = comparison.rename(columns={"CV R?": "CV R2"})
final_comparison = pd.DataFrame([
    {"Model": "Random Forest", "Stage": "Baseline", **baseline_results.loc["Random Forest"].to_dict()},
    {"Model": "Random Forest", "Stage": "Feature Engineered", **engineered_results.loc["Random Forest"].to_dict()},
    {"Model": "Random Forest", "Stage": "Tuned", **{k: v for k, v in tuned_rf_result.items() if k != "Model"}},
    {"Model": "XGBoost", "Stage": "Baseline", **baseline_results.loc["XGBoost"].to_dict()},
    {"Model": "XGBoost", "Stage": "Feature Engineered", **engineered_results.loc["XGBoost"].to_dict()},
    {"Model": "XGBoost", "Stage": "Tuned", **{k: v for k, v in tuned_xgb_result.items() if k != "Model"}},
])[["Model", "Stage", "CV R2", "CV MAE", "CV RMSE"]]
print(final_comparison.round(3).to_string(index=False))

best_row = final_comparison.loc[final_comparison["CV RMSE"].idxmin()]
if best_row["Model"] == "Random Forest":
    final_pipeline = make_engineered_pipeline(RandomForestRegressor(
        random_state=42, n_jobs=-1,
        n_estimators=rf_search.best_params_["model__n_estimators"],
        max_depth=rf_search.best_params_["model__max_depth"],
        min_samples_split=rf_search.best_params_["model__min_samples_split"],
        min_samples_leaf=rf_search.best_params_["model__min_samples_leaf"],
    ))
else:
    final_pipeline = make_engineered_pipeline(XGBRegressor(
        random_state=42, n_jobs=-1,
        n_estimators=xgb_search.best_params_["model__n_estimators"],
        learning_rate=xgb_search.best_params_["model__learning_rate"],
        max_depth=xgb_search.best_params_["model__max_depth"],
    ))
final_pipeline.fit(raw_X_train, y_train)

model_path = PROJECT_ROOT / "models" / "house_price_pipeline.joblib"
model_path.parent.mkdir(exist_ok=True)
joblib.dump(final_pipeline, model_path)
print(f"Selected final model: {best_row['Model']} ({best_row['Stage']})")
print(f"Saved pipeline: {model_path}")
```
